# Data visualization

```python
diagnosis_counts = data['diagnosis'].value_counts()

labels = ['Benign (B)', 'Malignant (M)']

plt.figure(figsize=(6,6))
plt.pie(
    diagnosis_counts,
    labels=labels,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Breast Cancer Diagnosis Distribution')
plt.show()
plt.figure(figsize=(6, 6))
ax = sns.countplot(data=data, x='diagnosis')

total = len(data)

for p in ax.patches:
    count = int(p.get_height())
    percentage = 100 * count / total

    ax.annotate(
        f'{count}\n{percentage:.1f}%',
        (p.get_x() + p.get_width()/2, count),
        ha='center',
        va='bottom'
    )

plt.title('Diagnosis Distribution')
plt.xlabel('Diagnosis')
plt.ylabel('Count')
plt.show()
```

# Data splitting

```python
X = data.drop('diagnosis', axis=1)
y = data['diagnosis']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
print(f'X_train.shape = {X_train.shape}')
print(f'y_train.shape = {y_train.shape}')
print(f'X_test.shape = {X_test.shape}')
print(f'y_test.shape = {y_test.shape}')
```

```python
from sklearn.preprocessing import StandardScaler
```

# Normalization

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

# ML Models

# Logistic Regression

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()
lr.fit(X_train, y_train)
```

```python
acc_lr = round(lr.score(X_train, y_train)*100, 2)
acc_lr
```

```python
y_pred_lr = lr.predict(X_test)
```

```python
sns.heatmap(confusion_matrix(y_test, y_pred_lr), annot=True, fmt='d', cmap='Reds')
```

```python
print(classification_report(y_test, y_pred_lr))
```

# Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier

dt = DecisionTreeClassifier(criterion='entropy', max_depth=3)
dt.fit(X_train, y_train)
```

```python
acc_dt = round(dt.score(X_train, y_train)*100, 2)
acc_dt
```

```python
y_pred_dt = dt.predict(X_test)
```

```python
sns.heatmap(confusion_matrix(y_test, y_pred_dt), annot=True, fmt='d', cmap='Reds')
```

```python
print(classification_report(y_test, y_pred_dt))
```

# Naive Bayes

```python
from sklearn.naive_bayes import GaussianNB

nb = GaussianNB()
nb.fit(X_train, y_train)
```

```python
acc_nb = round(nb.score(X_train, y_train)*100, 2)
acc_nb
```

```python
y_pred_nb = nb.predict(X_test)
```

```python
sns.heatmap(confusion_matrix(y_test, y_pred_nb), annot=True, fmt='d', cmap='Reds')
```

```python
print(classification_report(y_test, y_pred_nb))
```

# Support Vector Machine

```python
from sklearn.svm import SVC

svc = SVC(kernel='sigmoid', gamma=0.01)
svc.fit(X_train, y_train)
```

```python
acc_svc = round(svc.score(X_train, y_train)*100, 2)
acc_svc
```

```python
y_pred_svc = svc.predict(X_test)
```

```python
sns.heatmap(confusion_matrix(y_test, y_pred_svc), annot=True, fmt='d', cmap='Reds')
```

```python
print(classification_report(y_test, y_pred_svc))
```

# KNN

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=6)
knn.fit(X_train, y_train)
```

```python
acc_knn = round(knn.score(X_train, y_train)*100, 2)
acc_knn
```

```python
y_pred_knn = knn.predict(X_test)
```

```python
sns.heatmap(confusion_matrix(y_test, y_pred_knn), annot=True, fmt='d', cmap='Reds')
```

```python
print(classification_report(y_test, y_pred_knn))
```

# Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=3)
rf.fit(X_train, y_train)
```

```python
acc_rf = round(rf.score(X_train, y_train)*100, 2)
acc_rf
```

```python
y_pred_rf = rf.predict(X_test)
```

```python
sns.heatmap(confusion_matrix(y_test, y_pred_rf), annot=True, fmt='d', cmap='Reds')
```

```python
print(classification_report(y_test, y_pred_rf))
```
